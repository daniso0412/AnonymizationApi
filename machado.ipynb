## 6. Preparação dos dados

O texto será convertido em uma representação numérica adequada para o treinamento do modelo.

O corpus será dividido em dois conjuntos:

- treinamento;
- validação.

O conjunto de treinamento será utilizado para atualizar os parâmetros do modelo, enquanto o conjunto de validação será utilizado para acompanhar sua capacidade de generalização.

```python
%%writefile /content/nanoGPT/data/machado/prepare.py

import os
import pickle
import numpy as np

# ----------------------------------------------------------------------
# Carrega o texto
# ----------------------------------------------------------------------

input_file_path = os.path.join(os.path.dirname(__file__), 'input.txt')

with open(input_file_path, 'r', encoding='utf-8') as f:
    data = f.read()

print(f"Quantidade de caracteres: {len(data):,}")

# ----------------------------------------------------------------------
# Cria o vocabulário
# ----------------------------------------------------------------------

chars = sorted(list(set(data)))
vocab_size = len(chars)

print(f"Vocabulário: {vocab_size} caracteres")

stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for i, ch in enumerate(chars)}

def encode(s):
    return [stoi[c] for c in s]

def decode(l):
    return ''.join([itos[i] for i in l])

# ----------------------------------------------------------------------
# Divide em treino e validação
# ----------------------------------------------------------------------

n = len(data)

train_data = data[:int(n * 0.9)]
val_data = data[int(n * 0.9):]

train_ids = np.array(encode(train_data), dtype=np.uint16)
val_ids = np.array(encode(val_data), dtype=np.uint16)

# ----------------------------------------------------------------------
# Salva os arquivos
# ----------------------------------------------------------------------

train_ids.tofile(os.path.join(os.path.dirname(__file__), 'train.bin'))
val_ids.tofile(os.path.join(os.path.dirname(__file__), 'val.bin'))

meta = {
    'vocab_size': vocab_size,
    'itos': itos,
    'stoi': stoi,
}

with open(os.path.join(os.path.dirname(__file__), 'meta.pkl'), 'wb') as f:
    pickle.dump(meta, f)

print(f"Treino: {len(train_ids):,} tokens")
print(f"Validação: {len(val_ids):,} tokens")
print(f"Vocabulário: {vocab_size}")
print("Dataset preparado!")
```

## 12. Comparação

12.1 Pegando os valores do loss

```python
%cd /content/nanoGPT
!find out-machado-* -maxdepth 1 -type f 2>/dev/null
```

```python
from pathlib import Path
import re
import pandas as pd

experimentos = [
    ("2500", "out-machado-2500/training.log"),
    ("5000", "out-machado-5000/training.log"),
    ("10000", "out-machado-10000/training.log"),
]

resultados = []

for nome, arquivo in experimentos:
    texto = Path(arquivo).read_text()

    matches = re.findall(
        r"step\s+(\d+): train loss ([0-9.]+), val loss ([0-9.]+)",
        texto
    )

    for step, train_loss, val_loss in matches:
        resultados.append({
            "experimento": nome,
            "step": int(step),
            "train_loss": float(train_loss),
            "val_loss": float(val_loss)
        })

df = pd.DataFrame(resultados)

df.head()
```

```python
resultado_final = (
    df.sort_values("step")
      .groupby("experimento")
      .tail(1)
)

resultado_final
```

12.2 Menor perda de validação

```python
melhor_val = (
    df.loc[df.groupby("experimento")["val_loss"].idxmin()]
      .sort_values("experimento")
)

melhor_val
```

12.3 Grafico de validação e treinamento

```python
import matplotlib.pyplot as plt

for experimento in df["experimento"].unique():
    dados = df[df["experimento"] == experimento]

    plt.plot(
        dados["step"],
        dados["val_loss"],
        label=f"{experimento} iterações"
    )

plt.xlabel("Iterações")
plt.ylabel("Validation Loss")
plt.title("Validation Loss durante o treinamento")
plt.legend()
plt.grid(True)
plt.show()
```

```python
for experimento in df["experimento"].unique():
    dados = df[df["experimento"] == experimento]

    plt.plot(
        dados["step"],
        dados["train_loss"],
        label=f"{experimento} iterações"
    )

plt.xlabel("Iterações")
plt.ylabel("Train Loss")
plt.title("Train Loss durante o treinamento")
plt.legend()
plt.grid(True)
plt.show()
```
